# Data Filtering

**Testing:**
- After filtering, expect 1,000-3,000 4-year institutions in each dataset
- Verify filtering logic correctly identifies bachelor's-granting institutions
- Document how many institutions are excluded and why


```python
# Standardize institution ID column names
print("Standardizing institution ID column names...")

# Affordability data - rename to 'unit_id'
df_affordability = df_affordability.rename(columns={'Unit ID': 'unit_id'})

# College data - rename to 'unit_id'
df_college = df_college.rename(columns={
    'UNIQUE_IDENTIFICATION_NUMBER_OF_THE_INSTITUTION': 'unit_id'
})

print(f"✓ Affordability data: 'unit_id' column created")
print(f"✓ College data: 'unit_id' column created")
print(f"\nAffordability unit_id type: {df_affordability['unit_id'].dtype}")
print(f"College unit_id type: {df_college['unit_id'].dtype}")
```

```python
# Filter Affordability data to 4-year institutions
print("Filtering Affordability data to 4-year institutions...")
print(f"Before filtering: {len(df_affordability):,} institutions")

# Check available columns for filtering
if 'Highest Degree Offered Name' in df_affordability.columns:
    print("\nDegree levels in Affordability data:")
    print(df_affordability['Highest Degree Offered Name'].value_counts())
    
    # Filter for Bachelor's or higher
    df_affordability_4yr = df_affordability[
        df_affordability['Highest Degree Offered Name'].isin([
            "Bachelor's degree",
            "Master's degree", 
            "Doctor's degree - research/scholarship",
            "Doctor's degree -  professional practice",
            "Doctor's degree - research/scholarship and professional practice"
        ])
    ].copy()
elif 'Institution Level Name' in df_affordability.columns:
    print("\nInstitution levels in Affordability data:")
    print(df_affordability['Institution Level Name'].value_counts())
    
    df_affordability_4yr = df_affordability[
        df_affordability['Institution Level Name'] == 'Four or more years'
    ].copy()
else:
    print("Warning: Could not find filtering column. Using all data.")
    df_affordability_4yr = df_affordability.copy()

print(f"After filtering: {len(df_affordability_4yr):,} institutions")
print(f"Removed: {len(df_affordability) - len(df_affordability_4yr):,} institutions")
```

```python
# Filter College Results data to bachelor's-granting institutions
print("Filtering College Results data to bachelor's-granting institutions...")
print(f"Before filtering: {len(df_college):,} institutions")

# Check for degree level indicators
if 'Degree Level' in df_college.columns:
    print("\nDegree levels in College Results data:")
    print(df_college['Degree Level'].value_counts())
    
    df_college_4yr = df_college[
        df_college['Degree Level'] == 1  # 1 typically indicates bachelor's or higher
    ].copy()
elif 'Number of Bachelor Degrees Grand Total' in df_college.columns:
    # Filter for institutions that award bachelor's degrees
    df_college_4yr = df_college[
        df_college['Number of Bachelor Degrees Grand Total'].notna() & 
        (df_college['Number of Bachelor Degrees Grand Total'] > 0)
    ].copy()
else:
    print("Warning: Could not find filtering column. Using all data.")
    df_college_4yr = df_college.copy()

print(f"After filtering: {len(df_college_4yr):,} institutions")
print(f"Removed: {len(df_college) - len(df_college_4yr):,} institutions")
```

# Data Merging

**Testing:**
- Merge should keep 500-2,000 institutions
- Document merge rate (matched / total)
- Check for duplicate unit_ids
- Verify merge produces expected column count


```python
# Merge datasets on unit_id (inner join)
print("Merging datasets on unit_id...")
print("=" * 80)

# Check for duplicates before merging
print(f"\nAffordability data:")
print(f"  Total institutions: {len(df_affordability_4yr):,}")
print(f"  Unique unit_ids: {df_affordability_4yr['unit_id'].nunique():,}")
print(f"  Duplicates: {len(df_affordability_4yr) - df_affordability_4yr['unit_id'].nunique()}")

print(f"\nCollege Results data:")
print(f"  Total institutions: {len(df_college_4yr):,}")
print(f"  Unique unit_ids: {df_college_4yr['unit_id'].nunique():,}")
print(f"  Duplicates: {len(df_college_4yr) - df_college_4yr['unit_id'].nunique()}")

# Perform inner join
df_merged = pd.merge(
    df_affordability_4yr,
    df_college_4yr,
    on='unit_id',
    how='inner',
    suffixes=('_affordability', '_college')
)

# Calculate merge statistics
n_affordability = len(df_affordability_4yr)
n_college = len(df_college_4yr)
n_merged = len(df_merged)

merge_rate_aff = (n_merged / n_affordability * 100) if n_affordability > 0 else 0
merge_rate_college = (n_merged / n_college * 100) if n_college > 0 else 0

print("\n" + "=" * 80)
print("MERGE RESULTS")
print("=" * 80)
print(f"  From Affordability: {n_affordability:,} institutions")
print(f"  From College Results: {n_college:,} institutions")
print(f"  Merged (matched): {n_merged:,} institutions")
print(f"\n  Merge rate from Affordability: {merge_rate_aff:.1f}%")
print(f"  Merge rate from College Results: {merge_rate_college:.1f}%")
print(f"\n  Total columns in merged dataset: {df_merged.shape[1]}")
print("=" * 80)

# Show sample of merged data
print(f"\nMerged data sample:")
df_merged.head()
```

# Missing Data Analysis

**Testing:**
- Identify key variables with >20% missing
- Create missingness heatmap
- Document patterns (e.g., do small institutions have more missing data?)
- Prioritize critical variables for analysis


```python
# Calculate missing percentage for all columns
print("Calculating missing data statistics...")

missing_stats = pd.DataFrame({
    'column': df_merged.columns,
    'missing_count': df_merged.isnull().sum().values,
    'missing_percent': (df_merged.isnull().sum() / len(df_merged) * 100).values
}).sort_values('missing_percent', ascending=False)

print(f"\nTop 20 columns with most missing data:")
print(missing_stats.head(20))

# Visualize missingness
import missingno as msno

# Select subset of columns for visualization (too many to show all)
cols_to_viz = missing_stats.head(30)['column'].tolist()

fig, ax = plt.subplots(figsize=(12, 6))
msno.matrix(df_merged[cols_to_viz], ax=ax, sparkline=False)
plt.title("Missing Data Pattern (Top 30 Columns)")
plt.tight_layout()
plt.show()

# Summary statistics
print(f"\n{'='*80}")
print("MISSING DATA SUMMARY")
print(f"{'='*80}")
print(f"Total columns: {len(df_merged.columns)}")
print(f"Columns with NO missing: {(missing_stats['missing_percent'] == 0).sum()}")
print(f"Columns with <10% missing: {(missing_stats['missing_percent'] < 10).sum()}")
print(f"Columns with 10-50% missing: {((missing_stats['missing_percent'] >= 10) & (missing_stats['missing_percent'] < 50)).sum()}")
print(f"Columns with >50% missing: {(missing_stats['missing_percent'] >= 50).sum()}")
print(f"{'='*80}")
```

```python
# Identify critical variables and check their missingness
print("Checking missingness in CRITICAL variables...")
print("=" * 80)

critical_vars = {
    'Treatment': ['Affordability Gap (net price minus income earned working 10 hrs at min wage)', 'Net Price'],
    'Outcomes': [
        'Median Earnings of Students Working and Not Enrolled 10 Years After Entry',
        "Bachelor's Degree Graduation Rate Bachelor Degree Within 6 Years - Total"
    ],
    'Selectivity': [
        'Total Percent of Applicants Admitted',
        'SAT Evidence Based Reading and Writing - 25th Percentile Score',
        'ACT English - 25th Percentile Score'
    ],
    'Demographics': [
        "Percent of First-Time, Full-Time Undergraduates Awarded Pell Grants",
        'Percent of Bachelor Degrees White Total',
        'Percent of Bachelor Degrees Black or African American Total',
        'Percent of Bachelor Degrees Latino Total'
    ],
    'Resources': [
        'Instructional Expenses Per FTE',
        'Endowment Assets FASB per Student'
    ],
    'Institution': [
        'Sector of Institution',
        'State of Institution',
        'Institution Size Category'
    ],
    'MSI': ['HBCU', 'HSI', 'TCU']
}

for category, vars in critical_vars.items():
    print(f"\n{category}:")
    for var in vars:
        if var in df_merged.columns:
            missing_pct = (df_merged[var].isnull().sum() / len(df_merged) * 100)
            status = "✓" if missing_pct < 20 else "⚠" if missing_pct < 50 else "✗"
            print(f"  {status} {var[:60]:<60} Missing: {missing_pct:5.1f}%")
        else:
            print(f"  ✗ {var[:60]:<60} NOT FOUND")

print("=" * 80)
```

# Data Cleaning

**Testing:**
- After listwise deletion, final sample should be 500+ institutions
- Verify no missing values remain in critical columns
- Document which institutions are excluded and why
- Check if excluded institutions differ systematically from included ones


```python
# Apply listwise deletion for rows missing critical variables
print("Applying listwise deletion for critical variables...")
print("=" * 80)

# Define absolutely required variables (must be present for analysis)
required_vars = [
    'Affordability Gap (net price minus income earned working 10 hrs at min wage)',
    'Median Earnings of Students Working and Not Enrolled 10 Years After Entry',
    "Bachelor's Degree Graduation Rate Bachelor Degree Within 6 Years - Total"
]

print(f"\nRequired variables for analysis:")
for var in required_vars:
    if var in df_merged.columns:
        print(f"  ✓ {var}")
    else:
        print(f"  ✗ {var} - NOT FOUND (will try to find alternative)")

# Filter out rows with missing required variables
df_clean = df_merged.copy()
initial_n = len(df_clean)

for var in required_vars:
    if var in df_clean.columns:
        before = len(df_clean)
        df_clean = df_clean[df_clean[var].notna()].copy()
        after = len(df_clean)
        removed = before - after
        print(f"\n  Removed {removed:,} institutions missing '{var[:50]}...'")
        print(f"    Remaining: {after:,}")

final_n = len(df_clean)
removed_total = initial_n - final_n

print(f"\n{'='*80}")
print(f"CLEANING RESULTS")
print(f"{'='*80}")
print(f"  Before cleaning: {initial_n:,} institutions")
print(f"  After cleaning: {final_n:,} institutions")
print(f"  Removed: {removed_total:,} institutions ({removed_total/initial_n*100:.1f}%)")
print(f"{'='*80}")

# Verify completeness
print(f"\nVerifying completeness of required variables...")
for var in required_vars:
    if var in df_clean.columns:
        missing = df_clean[var].isnull().sum()
        print(f"  {var[:50]:<50} Missing: {missing}")

# Check if we met our minimum sample size goal
if final_n >= 500:
    print(f"\n✓ SUCCESS: Final sample of {final_n:,} institutions exceeds minimum (500)")
else:
    print(f"\n⚠ WARNING: Final sample of {final_n:,} institutions is below minimum (500)")
```
